```python
import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
GRN_ROOT = './data'
```

```python
male_GRN = pd.read_csv(f'{GRN_ROOT}/100per_act_inferred_GRN_MALE.csv', index_col = 0)
female_GRN = pd.read_csv(f'{GRN_ROOT}/100per_act_inferred_GRN_FEMALE.csv', index_col=0)
```

```python
male_TFs = set(male_GRN['TF'])
male_TGs = set(male_GRN['Gene'])
```

```python
male_TF_TG_pairs = [0] * male_GRN.shape[0]
for row_index in range(0, male_GRN.shape[0]):
    male_TF_TG_pairs[row_index] = f'{male_GRN.iloc[row_index, 0]}_{male_GRN.iloc[row_index,1]}'
```

```python
female_TF_TG_pairs = [0] * female_GRN.shape[0]
for row_index in range(0, female_GRN.shape[0]):
    female_TF_TG_pairs[row_index] = f'{female_GRN.iloc[row_index, 0]}_{female_GRN.iloc[row_index,1]}'
```

```python
#make the pairs sets for easy comparisons
female_pairs = set(female_TF_TG_pairs)
male_pairs = set(male_TF_TG_pairs)
```

```python
#get sex-specific pairs
female_specific_pairs = female_pairs - male_pairs
male_specific_pairs = male_pairs - female_pairs
```

```python
female_specific_TFs = [0] * len(female_specific_pairs)
female_specific_TGs = [0] * len(female_specific_pairs)
for pair_index in range(0, len(female_specific_pairs)):
    female_specific_TFs[pair_index] = list(female_specific_pairs)[pair_index].split('_')[0]
    female_specific_TGs[pair_index] = list(female_specific_pairs)[pair_index].split('_')[1]

female_specific_genes = female_specific_TFs + female_specific_TGs
```

```python
with open('./data/female_queary.txt', 'w+') as f:
    for gene in female_specific_genes:
        f.write(f'{gene}\n')
f.close
```

```python
male_specific_TFs = [0] * len(male_specific_pairs)
male_specific_TGs = [0] * len(male_specific_pairs)
for pair_index in range(0, len(male_specific_pairs)):
    male_specific_TFs[pair_index] = list(male_specific_pairs)[pair_index].split('_')[0]
    male_specific_TGs[pair_index] = list(male_specific_pairs)[pair_index].split('_')[1]

male_specific_genes = male_specific_TFs + male_specific_TGs
```

```python
with open('./data/male_queary.txt', 'w+') as f:
    for gene in male_specific_genes:
        f.write(f'{gene}\n')
f.close
```
