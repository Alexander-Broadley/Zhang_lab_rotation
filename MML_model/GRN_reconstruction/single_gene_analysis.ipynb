```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
diff_reg_results = pd.read_csv('/home/alexanderb/Zhang_lab/MML_model/GRN_reconstruction/data/GW_diff_regulatorts_0.1_highAcc.csv', index_col = 0)
```

```python
diff_reg_results = diff_reg_results.drop_duplicates()
```

```python
print(len(set(diff_reg_results['TF'])))
print(len(set(diff_reg_results['TG'])))
```

```python
fig, ax = plt.subplots(ncols = 1, figsize = (10, 10))
sns.scatterplot(diff_reg_results, x = 'abs_diff', y = 'TF_exp_difference', ax = ax, s = 8, alpha = 0.8, color = 'darkblue')
plt.xlabel('Absolute Difference in TF Normalised Expression Contribution', fontsize = 15)
plt.ylabel('Absolute Difference in Mean log TF Expression', fontsize = 15)
ax.tick_params(axis='both', which='major', labelsize=15)
plt.savefig('./figures/cont_v_exp_scatter', bbox_inches = 'tight', dpi = 300)
```
